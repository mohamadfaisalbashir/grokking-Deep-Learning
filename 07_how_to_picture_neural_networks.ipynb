## **2. Correlation Summarization**

**This idea is the key to moving forward to more advanced networks without losing one's footing.**

### **2.1 The Overall Statement**

> **Correlation summarization:** a neural network tries to find direct and indirect correlation between an input layer and an output layer, which are determined by the input and output datasets respectively.

This is what all neural networks do when viewed from a "10,000-foot" level. Since a neural network is really a series of matrices connected by layers, the author next zooms in a little and asks what any single weight matrix is doing.

### **2.2 The Local View**

> **Local correlation summarization:** any given set of weights optimizes itself to learn how to correlate its input layer with what the output layer says it should be.

With only two layers (input and output), the weight matrix knows what the output layer should be from the output dataset, and it simply looks for correlation between the input and output datasets. With several layers the situation is more nuanced.

### **2.3 The Global View and Backpropagation**

> **Global correlation summarization:** what an earlier layer should be can be determined by taking what a later layer says it should be and multiplying it by the weights between them. In this way later layers tell earlier layers what kind of signal they need in order to find correlation with the output. This cross-communication is called **backpropagation**.

Once global correlation has taught every layer what it should be, local correlation can optimize the weights locally. The author uses a "game of telephone" as an analogy:

- a neuron in the final layer announces that it needs to be a little higher, and asks every neuron of the preceding layer to send it a higher signal;
- those neurons pass the same request on to the layer before them;
- at the end of the game, every layer knows which of its neurons should be higher and which lower, and the local correlation summarization takes over and **updates the weights accordingly**.

**The dimensions of the matrices are determined by the layers around them.**

The previous section hides a pattern: the number of rows and columns of every matrix is directly related to the dimensionality of the layers before and after it. The picture can therefore be simplified once more, by dropping the shape annotations and the circles.

All the information needed to build the network is still present. For example, `weights_0_1` must be a **(3 × 4) matrix**: the previous layer (`layer_0`) has three dimensions and the next layer (`layer_1`) has four, so the matrix needs exactly one weight connecting each node of `layer_0` to each node of `layer_1`.

### **5.1 What the Network Does**

Seen through the correlation summarization, this network simply **adjusts its weights to find correlation between `layer_0` and `layer_2`**, using all the methods discussed so far in the book. But the configuration of weights and layers placed between the input and output layers strongly affects **whether** the network finds correlation, and/or **how fast** it does so.

### **5.2 Architecture**

The particular configuration of layers and weights in a network is called its **architecture**, and the majority of the rest of the book discusses the advantages and disadvantages of various architectures. As the correlation summarization reminds us, the network adjusts weights to find correlation between the input and output layers, sometimes even inventing correlation in the hidden layers. Different architectures **channel signal** so that correlation becomes easier to discover.

> Good neural architectures channel signal so that correlation is easy to discover. Great architectures also filter noise, which helps prevent overfitting.

Much of the research on neural networks is about finding new architectures that find correlation faster and generalize better to unseen data, and most of the remainder of the book is devoted to new architectures.

## **6. Let's See This Network Predict**

**Data from the streetlight example flows through the system.**

The figure below follows a single datapoint from the streetlight dataset through the network in three steps (numbered 1 to 3 in the book).

> A prediction in three steps: (1) layer_0 is set to [1, 0, 1]; (2) layer_1 becomes [.5, .2, 0, .9]; (3) layer_2 becomes .9
>
> <img src="pictures/chapter 7/picture 7.5.png" alt="Figure 7.5" width="300"/>

1. **Set the input.** A single datapoint of the streetlight dataset is selected, and `layer_0` is set to the correct values: **1, 0, 1**.
2. **Compute the hidden layer.** Four different weighted sums of `layer_0` are computed by `weights_0_1`. This operation is called **vector-matrix multiplication**. The four results are placed in the four positions of `layer_1` and passed through the `relu` function, which sets negative values to 0. The resulting values are **.5, .2, 0, and .9**; the third value (from the left) would have been negative, but `relu` set it to 0.
3. **Compute the prediction.** A final vector-matrix multiplication, now with `weights_1_2`, turns `layer_1` into a single number. This is the network's final prediction: **.9**. (The book describes this last step as a weighted average of `layer_1`.)

### **6.1 Review: Vector-Matrix Multiplication**

Vector-matrix multiplication performs **several weighted sums of one vector**. The matrix must have as many **rows** as the vector has values, and each **column** of the matrix performs its own, unique weighted sum. A matrix with four columns therefore produces four weighted sums, and the weightings used in each sum are given by the values in the matrix. In this network, the (3 × 4) matrix `weights_0_1` turns the three values of `layer_0` into four weighted sums, and the (4 × 1) matrix `weights_1_2` turns the four values of `layer_1` into one.

## **7. Visualizing Using Letters Instead of Pictures**

**All these pictures and detailed explanations amount to a simple piece of algebra.**

Just as simpler pictures were defined for matrices and vectors, the same can be done with letters.

### **7.1 Letters for Matrices and Vectors**

- **A matrix** is written as a **capital letter**. The author chooses $W$ (for "weights"), because it is easy to remember. The small subscript, such as the 0 in $W_0$, signals that it is one of several $W$s (this network has two) and lets all weight matrices share the letter while remaining distinguishable. Any capital letter would work; the notation is personal, so it should be easy for the reader to remember.
- **A vector** is written as a **lowercase letter**. The author chooses $l$, because the vectors of this network are layers. Subscripts again number them ($l_0$ is read "l-zero"), which avoids inventing a new letter for every layer. There is no wrong answer here either.

> A matrix is named with a capital letter (here W0) and a vector with a lowercase letter (here l0)
>
> <img src="pictures/chapter 7/picture 7.6.png" alt="Figure 7.6" width="450"/>

### **7.2 Naming Every Piece of the Network**

Applied to the whole network, the letters point to their bricks as follows:

| Brick | Letter |
|---|---|
| `layer_0` | $l_0$ |
| `weights_0_1` | $W_0$ |
| `layer_1` | $l_1$ |
| `weights_1_2` | $W_1$ |
| `layer_2` | $l_2$ |

> Every brick of the network labelled with its variable: l0, W0, l1, W1, and l2

Defining the variables does not yet show how they relate to each other. The next section combines them through vector-matrix multiplication.

## **8. Linking the Variables**

**Letters can be combined to express functions and operations.**

### **8.1 Multiplication**

To say that two letters are multiplied, they are simply written next to each other (here, vector-matrix multiplication):

| Algebra | Translation |
|---|---|
| $l_0 W_0$ | Take the layer 0 vector and perform vector-matrix multiplication with weight matrix 0. |
| $l_1 W_1$ | Take the layer 1 vector and perform vector-matrix multiplication with weight matrix 1. |

### **8.2 Adding Functions Such as `relu`**

Arbitrary functions, such as `relu`, can be included with a notation that looks almost exactly like the Python code:

| Algebra | Translation |
|---|---|
| $l_1 = \text{relu}(l_0 W_0)$ | To create the layer 1 vector, multiply the layer 0 vector by weight matrix 0 (vector-matrix multiplication), and then apply `relu` to the output, setting all negative numbers to 0. |
| $l_2 = l_1 W_1$ | To create the layer 2 vector, multiply the layer 1 vector by weight matrix 1 (vector-matrix multiplication). |

### **8.3 The Whole Network in One Expression**

The algebra of layer 2 contains layer 1 as an input variable, so the two expressions can be chained, and the **entire network** is represented by one formula:

$$l_2 = \text{relu}(l_0 W_0)\,W_1$$

All the logic of the forward propagation step is contained in this single formula. Note that it silently assumes that the vectors and matrices have the right dimensions.

## **9. Everything Side by Side**

**The picture, the algebra, and the Python code in one place.**

The author considers little commentary necessary here: the reader should simply look at forward propagation in these four different ways (the Python line, the algebraic formula, the node-and-arrow drawing, and the brick drawing) in order to understand it, and the architecture, from several perspectives at once.

The Python line printed in the book is:

### **10.1 Key Takeaway**

> Good neural architectures channel signal so that correlation is easy to discover. Great architectures also filter noise to help prevent overfitting.

A neural architecture controls **how signal flows through a network**, and the way it is built affects which correlations the network can detect. The aim is to create architectures that **maximize the network's ability to focus on areas where meaningful correlation exists**, and **minimize its ability to focus on areas that contain only noise**.

Different datasets and domains have different characteristics. Image data, for instance, contains different kinds of signal and noise than text data. Although neural networks can be used in many situations, different architectures are better suited to different problems, because each is better at locating certain kinds of correlation. For the next few chapters, the book therefore explores how to modify neural networks so that they find the specific correlation one is looking for.
