## **1. What Is Deep Learning?**

**Deep learning is a subset of methods for machine learning.**

Machine learning is a field dedicated to the study and development of machines that can learn, sometimes with the long-term goal of eventually attaining general artificial intelligence. Deep learning sits inside this field.

In industry, deep learning is used to solve practical tasks in a variety of areas:

- **Computer vision** (images)
- **Natural language processing** (text)
- **Automatic speech recognition** (audio)

In short, deep learning is a subset of *methods* in the machine learning toolbox, and it primarily uses **artificial neural networks**, a class of algorithms loosely inspired by the human brain.

## **2. What Is Machine Learning?**

A field of study that gives computers the ability to learn without being explicitly programmed. This definition is attributed to Arthur Samuel.

Machine learning is a subfield of computer science in which machines *learn* to perform tasks that they were *not explicitly programmed* to do. In short, machines observe a pattern and attempt to imitate it, either directly or indirectly. The author summarizes this with a memorable analogy: **machine learning is approximately "monkey see, monkey do."**

> Machine learning is approximately monkey see, monkey do
>
> <img src="pictures/chapter 2/picture 2.2.png" alt="Figure 2.2" width="600"/>

The distinction between direct and indirect imitation parallels the two main types of machine learning:

- **Supervised learning** is the *direct* imitation of a pattern between two datasets.
- **Unsupervised learning** is the *indirect* imitation of a pattern.

Supervised learning always attempts to take an **input dataset** and transform it into an **output dataset**. This can be an extremely powerful and useful capability. The following examples show input datasets and the output datasets they are used to predict:

| Input dataset | Output dataset |
|---------------|----------------|
| Pixels of an image | Presence or absence of a cat |
| Movies you have liked | Movies you may like |
| Someone's words | Whether they are happy or sad |
| Weather sensor data | Probability of rain |
| Car engine sensors | Optimal tuning settings |
| News data | Tomorrow's stock price |
| An input number | A number double its size |
| A raw audio file | A transcript of the audio |

All of these are supervised machine learning tasks. In each case, the algorithm imitates the pattern between the two datasets so that it can **use one dataset to predict the other**. Imagine having the ability to predict the output dataset given only the input dataset: such an ability would be profound.

## **3. Supervised Machine Learning**

**Supervised learning transforms datasets.**

Supervised learning is a method for transforming one dataset into another. For example, suppose there is a dataset called *Monday Stock Prices*, which records the price of every stock on every Monday for the past 10 years, and a second dataset called *Tuesday Stock Prices*, recorded over the same period. A supervised learning algorithm might try to use one to predict the other.

> Supervised learning maps Monday stock prices to Tuesday stock prices
>
> <img src="pictures/chapter 2/picture 2.3.png" alt="Figure 2.3" width="600"/>

If the algorithm were successfully trained on 10 years of Mondays and Tuesdays, it could predict the stock price on any future Tuesday given the price on the immediately preceding Monday.

Supervised machine learning is the **bread and butter of applied artificial intelligence** (also known as narrow AI). It is useful for taking *what you know* as input and quickly transforming it into *what you want to know*. This allows supervised learning algorithms to extend human intelligence and capabilities in a seemingly endless number of ways.

> Supervised learning maps what you know to what you want to know
>

The majority of machine learning work results in training a supervised classifier of some kind. Even unsupervised learning is typically used to help develop an accurate supervised algorithm. For the rest of the book, the goal is to build algorithms that take input data that is observable, recordable, and therefore *knowable*, and transform it into valuable output data that requires logical analysis.

## **4. Unsupervised Machine Learning**

**Unsupervised learning groups your data.**

Unsupervised learning shares one property with supervised learning: it transforms one dataset into another. The difference is that the dataset it transforms into is *not previously known or understood*. There is no "right answer" for the model to duplicate. The algorithm is simply told to find patterns in the data and report on them.

**Clustering** is the standard example. Clustering transforms a sequence of datapoints into a sequence of **cluster labels**. If the algorithm learns 10 clusters, the labels are commonly the numbers 1 to 10, and each datapoint is assigned the number of the cluster to which it belongs. The dataset thus turns from a collection of datapoints into a collection of labels.

The labels are numbers because the algorithm does not say what the clusters *are*. It only reports that it has found some structure, in effect telling the scientist that there appear to be groups in the data.

> Unsupervised learning maps a list of datapoints to a list of cluster labels
>
> <img src="pictures/chapter 2/picture 2.5.png" alt="Figure 2.5" width="600"/>

The author offers a useful rule to hold onto: although there are many forms of unsupervised learning, **all forms of unsupervised learning can be viewed as a form of clustering**. This idea is developed further later in the book.

> Clustering the words puppies, pizza, kittens, hot dog, burger
>
> <img src="pictures/chapter 2/picture 2.6.png" alt="Figure 2.6" width="600"/>

In this example, the algorithm does not name the clusters, but the pattern can be inferred: cluster 1 corresponds to *cute* things and cluster 2 corresponds to *delicious* things. Later chapters show how other forms of unsupervised learning are also forms of clustering, and why these clusters are useful for supervised learning.

## **5. Parametric vs. Nonparametric Learning**

**Oversimplified: trial-and-error learning vs. counting and probability.**

The previous sections divided all machine learning algorithms into supervised and unsupervised. There is a second, independent way to divide the same algorithms: **parametric** and **nonparametric**. Conceptually, the "machine learning cloud" has two switches.

> The machine learning cloud with two switches
>
> <img src="pictures/chapter 2/picture 2.7.png" alt="Figure 2.7" width="600"/>

This yields four types of algorithms:

| | Parametric | Nonparametric |
|---|---|---|
| **Supervised** | Supervised parametric | Supervised nonparametric |
| **Unsupervised** | Unsupervised parametric | Unsupervised nonparametric |

The supervised/unsupervised axis concerns the **type of pattern** being learned. The parametric/nonparametric axis concerns **how the learning is stored** and, by extension, the **method of learning**. There is still some debate about the exact difference, but the formal definition used here is:

> A parametric model has a fixed number of parameters, whereas a nonparametric model's number of parameters is infinite (determined by data).

The author illustrates this with a **square peg and square hole** analogy:

- Some humans, such as babies, jam the peg into every hole until it fits somewhere. This is **parametric**: trial and error.
- A teenager may count the number of sides (four) and search for a hole with the same number. This is **nonparametric**: counting.

In general, parametric models tend to use trial and error, whereas nonparametric models tend to count.

## **6. Supervised Parametric Learning**

**Oversimplified: trial-and-error learning using knobs.**

Supervised parametric learning machines have a **fixed number of knobs**, which is the parametric part. Learning occurs by turning those knobs. Input data enters the machine, is processed according to the angle of each knob, and is transformed into a **prediction**.

> Data enters a machine with knobs and produces a prediction
>
> <img src="pictures/chapter 2/picture 2.8.png" alt="Figure 2.8" width="600"/>

Learning is accomplished by turning the knobs to different angles. Consider predicting the probability that the Red Sox will win the World Series. The model takes sports statistics, such as the win/loss record or the average number of toes per player, and produces a prediction (for example, a 98% chance). It then observes whether the team actually won. Once the truth is known, the learning algorithm **updates the knobs** so that the next prediction on the same or similar input data is more accurate.

- If the win/loss record was a good predictor, the model might turn that knob *up*.
- If the average number of toes was not a good predictor, the model might turn that knob *down*.

Several observations follow from this description:

- The entirety of what the model has learned is captured in the **positions of the knobs** at any given time.
- This type of model can be viewed as a **search algorithm**: it searches for the right knob configuration by trying one, adjusting, and retrying.
- Trial and error is not the formal definition of parametric learning, but it is a common property, with exceptions. When there is an arbitrary but fixed number of knobs, some searching is needed to find the optimal configuration. By contrast, nonparametric learning is often count-based and more or less adds new knobs when it finds something new to count.

Supervised parametric learning can be broken down into three steps.

### **Step 1: Predict**

Continuing the sports analogy, the first step is to gather sports statistics, pass them through the machine, and make a prediction about the probability that the Red Sox will win. In the book's example, the input includes the location (away), the opponent (Yankees), the number of toes (250), the number of players (25), and the number of fans (25,000), and the machine predicts 98%.

> Step 1: predict from sports statistics
>
> <img src="pictures/chapter 2/picture 2.9.png" alt="Figure 2.9" width="600"/>

### **Step 2: Compare to the Truth Pattern**

The second step compares the prediction with the pattern of interest, namely whether the Red Sox actually won. Unfortunately, they lost, so the comparison is:

`Pred: 98%  >  Truth: 0%`

This step recognizes that if the model had predicted 0%, it would have predicted the loss perfectly. Since the goal is an accurate machine, this leads to the third step.

### **Step 3: Learn the Pattern**

This step adjusts the knobs by studying both *how much* the model missed by (98%) and *what the input data was* at the time of prediction. It then turns the knobs to produce a more accurate prediction for that input.

In theory, the next time the model sees the same statistics, its prediction will be lower than 98%. Each knob represents the **prediction's sensitivity to a different type of input data**, and this sensitivity is exactly what changes when the model "learns."

> Adjusting sensitivity by turning knobs
>

## **7. Unsupervised Parametric Learning**

Unsupervised parametric learning uses a very similar approach. Since unsupervised learning is about grouping data, the parametric version uses knobs to group it. Usually there are several knobs **for each group**, and each group's knobs map the input data's **affinity** to that particular group. This is a high-level description, with exceptions and nuance.

The book illustrates this with a small dataset of home/away status and number of fans, which the author has divided into three clusters for the model to find (group 1, group 2, and group 3).

> Small dataset of home or away and number of fans
>
> <img src="pictures/chapter 2/picture 2.11.png" alt="Figure 2.11" width="600"/>

The first datapoint (home, 100k fans) is propagated through a trained unsupervised model. Each group has its own machine with a "# fans" knob and a "home/away" knob, and each machine outputs a number between 0 and 1 representing the **probability that the datapoint belongs to that group**. In the example, the datapoint maps most strongly to group 1.

> Group membership probabilities for one datapoint

There is considerable variety in how these models train and in their resulting properties. At a high level, however, they adjust their parameters so that the input data is transformed into the group or groups to which it belongs.

## **8. Nonparametric Learning**

**Oversimplified: counting-based methods.**

Nonparametric learning is a class of algorithms in which the **number of parameters is based on the data** instead of being predefined. This lends itself to methods that count in one way or another, increasing the number of parameters according to the number of items counted in the data.

In a supervised setting, for example, a nonparametric model might count how many times a particular color of streetlight causes cars to "go." After counting only a few examples, the model could predict that *middle* lights always (100%) cause cars to go, while *right* lights only sometimes (50%) do.

> Counting streetlight colors and whether cars go or stop
>
> <img src="pictures/chapter 2/picture 2.13.png" alt="Figure 2.13" width="600"/>

This model would have three parameters: three counts of how many times each colored light was on and cars went (possibly divided by the total number of observations). With five lights there would be five counts and therefore five parameters. What makes the model **nonparametric** is that the number of parameters changes with the data, here the number of lights. In contrast, a parametric model starts with a set number of parameters and, more importantly, can have more or fewer of them at the discretion of the scientist, regardless of the data.

Two clarifications are worth noting:

- **There is a gray area.** A parametric model may seem to have a knob for each input datapoint, since most parametric models must still have some form of input sized to the number of classes in the data. Even parametric algorithms are therefore somewhat influenced by the data, even if they do not explicitly count patterns.
- **"Parameters" is a generic term.** It refers only to the set of numbers used to model a pattern, with no limitation on how those numbers are used. Counts, weights, normalized variants of counts or weights, and correlation coefficients can all be parameters.

Deep learning is a class of **parametric** models. Nonparametric models are not discussed further in the book, although the author notes that they are an interesting and powerful class of algorithm.

## **9. Summary**

Chapter 2 goes one level deeper into the different flavors of machine learning. The key takeaways are:

- **Deep learning** is a subset of **machine learning**, which in turn overlaps with artificial intelligence. Deep learning primarily uses artificial neural networks.
- **Machine learning** gives computers the ability to learn without being explicitly programmed, and can be summarized as "monkey see, monkey do."
- Every machine learning algorithm is either **supervised** or **unsupervised**, and either **parametric** or **nonparametric**. These are four distinct groups.
- **Supervised learning** learns to predict one dataset given another. It is the bread and butter of applied AI, transforming *what you know* into *what you want to know*.
- **Unsupervised learning** generally groups a single dataset into clusters, and all of its forms can be viewed as clustering.
- **Parametric** algorithms have a fixed number of parameters and generally learn by trial and error (turning knobs). **Nonparametric** algorithms adjust their number of parameters based on the dataset and generally learn by counting.
- Supervised parametric learning follows three steps: **predict, compare to the truth pattern, and learn the pattern**.
- Deep learning uses neural networks to perform both supervised and unsupervised prediction.
